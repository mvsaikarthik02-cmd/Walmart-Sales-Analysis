```python
# Strip the '$' sign and remove any accidental whitespace
df['unit_price'] = df['unit_price'].astype(str).str.replace('$', '', regex=False).str.strip()

# Convert the column to numeric, turning unconvertible values into NaN (Not a Number)
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')

print("Sample unit prices after cleaning:\n", df['unit_price'].head())
```

```python
# Convert 'date' column to standard datetime format
# The raw data is in Day/Month/Year format (e.g., 05/01/19)
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%y', errors='coerce')
# If there are any stray rows with alternate formats, let pandas infer them
df['date'] = df['date'].fillna(pd.to_datetime(df['date'], errors='coerce'))
# Ensure 'time' format is standard HH:MM:SS
df['time'] = pd.to_datetime(df['time'], format='%H:%M:%S', errors='coerce').dt.time
print("\nUpdated Date & Time Data Types:")
print(df[['date', 'time']].dtypes)
```

```python
# Check if the missingness in both columns is identical for every row
are_rows_identical = (df['unit_price'].isnull() == df['quantity'].isnull()).all()

print(f"Are the missing values in the exact same rows? {are_rows_identical}")
```

```python
# Count rows where BOTH columns are missing
both_empty_count = (df['unit_price'].isnull() & df['quantity'].isnull()).sum()

print(f"Number of rows where both are empty: {both_empty_count}")
```

```python
df['unit_price'] = df.groupby('category')['unit_price'].transform(lambda x: x.fillna(x.median()))
median_quantity = df['quantity'].median()
df['quantity'] = df['quantity'].fillna(median_quantity).round(0).astype(int)
```

```python
#3.Feature Engineering
# --- 1. FINANCIAL FEATURES ---
# Calculate total sales revenue per transaction
df['total_revenue'] = (df['unit_price'] * df['quantity']).round(2)

# Calculate total profit per transaction based on profit_margin percentage
df['total_profit'] = (df['total_revenue'] * df['profit_margin']).round(2)

# --- 2. DATE-BASED FEATURES ---
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['month_name'] = df['date'].dt.strftime('%b') # e.g., Jan, Feb
df['day_of_week'] = df['date'].dt.strftime('%A') # e.g., Monday, Tuesday
df['is_weekend'] = df['date'].dt.dayofweek.isin([5, 6]).astype(int) # 1 if Sat/Sun, else 0

# --- 3. TIME-BASED FEATURES ---
# Extract hour from time string
df['hour'] = pd.to_datetime(df['time'], format='%H:%M:%S').dt.hour
# Bin hours into time of day categories
bins = [0, 12, 17, 24]
labels = ['Morning', 'Afternoon', 'Evening']
df['time_of_day'] = pd.cut(df['hour'], bins=bins, labels=labels, right=False)

print(df[['total_revenue', 'total_profit', 'month_name', 'day_of_week', 'is_weekend', 'time_of_day']].head())
```

```python
"""
EDA Questions
1.Find Top 10 Performing Cities by Total Sales
2.Find Top Categories By Total Sales
3.Find Peak Sales Hours
4.Find Peak Sales Days
5.Annual Yearly Sales
6.Annual Monthly Sales
7.Pay Method Distribution
8.Average Customer rating by category
9.Average Order Value by payment method
10.Find Peak Time of Day Sales
"""
```

```python
#Sales Analysis
#1.Top 10 Performing Cities by Total Sales
top_cities = df.groupby('City').agg(Total_Revenue=('total_revenue', 'sum'),Transactions=('invoice_id', 'count')).reset_index().sort_values(by='Total_Revenue', ascending=False).head(10)
print("\n=== TOP 10 CITIES BY REVENUE ===")
print(top_cities.to_string(index=False))
```

```python
# 2.Top Categories by Sales 
cat_analysis = df.groupby('category').agg(
    Total_Revenue=('total_revenue', 'sum'),
    Total_Units=('quantity', 'sum'),
    Total_Transactions=('invoice_id', 'count'),
    Avg_Unit_Price=('unit_price', 'mean'),
    Avg_Order_Value=('total_revenue', 'mean')
).reset_index()

cat_analysis['Revenue_Share_%'] = (cat_analysis['Total_Revenue'] / df['total_revenue'].sum()) * 100
cat_analysis = cat_analysis.sort_values(by='Total_Revenue', ascending=False)

print("=== CATEGORY SALES & REVENUE ANALYSIS ===")
print(cat_analysis.to_string(index=False))
```

```python
# 3. Peak Sales Hours
hourly_sales = df.groupby('hour')['total_revenue'].sum().reset_index()
hourly_sales.sort_values(by='total_revenue',ascending=True)
print(hourly_sales.to_string(index=False))
```

```python
#4.Peak sales on Days
timeday_sales = df.groupby('time_of_day')['total_revenue'].sum().reset_index()
timeday_sales.sort_values(by='total_revenue',ascending=True)
print(timeday_sales.to_string(index=False))
```

```python
#5.Peak Day of week sales
day_sales = df.groupby('day_of_week')['total_revenue'].sum().reset_index()
day_sales.sort_values(by='total_revenue',ascending=True)
print(day_sales.to_string(index=False))
```

```python
# 6.Annual totals
annual_sales = df.groupby('year')['total_revenue'].sum().reset_index()
print("=== Annual Total Sales ===")
print(annual_sales)

# 7.Average sales by month across 2020-2023
df_2020_2023 = df[df['year'].between(2019, 2023)]
avg_monthly_2020_2023 = df_2020_2023.groupby('month')['total_revenue'].mean().reset_index()
print("\n=== Average Monthly Sales (2019-2023) ===")
print(avg_monthly_2020_2023)
```

```python
#2.Customer Analysis
# 8. Payment Method Distribution
payment_counts = df['payment_method'].value_counts(normalize=True) * 100
print("=== Payment Method Usage (%) ===")
print(payment_counts)

# 9. Average Customer Rating by Category
avg_rating_cat = df.groupby('category')['rating'].agg(['mean', 'median', 'count']).reset_index()
print("\n=== Customer Rating by Category ===")
print(avg_rating_cat.sort_values(by='mean', ascending=False))

# 10. Average Order Value (AOV) by Payment Method
aov_payment = df.groupby('payment_method').agg(
    avg_spend=('total_revenue', 'mean'),
    avg_quantity=('quantity', 'mean'),
).reset_index()
print("\n=== Customer Behavior by Payment Method ===")
print(aov_payment)
```
